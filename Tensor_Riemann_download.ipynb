```python
import einsteinpy.symbolic as es
import sympy as sp
import numpy as np
from IPython.display import display
```

```python
T, x, y, z = sp.symbols('T x y z')

# Minkowski en cart
gmat = np.diag([1, -1, -1, -1])

g = es.MetricTensor(gmat, [T, x, y, z])
display(g.tensor())
```

```python
Gama = es.ChristoffelSymbols.from_metric(g)
display(Gama.tensor())
```

```python
T, r, teta, z = sp.symbols('T r teta z')

# Minkowski en cilindricas
gmat = np.diag([1, -1, -r**2, -1])
g = es.MetricTensor(gmat, [T, r, teta, z])
display(g.tensor())
```

```python
Gama = es.ChristoffelSymbols.from_metric(g)
display(Gama.tensor())
```

```python
Riemann = es.RiemannCurvatureTensor.from_metric(g)
display(Riemann.tensor())
```

### Ahora con la métrica de Newton

```python
# En coordenadas esféricas
T, r, teta, phi, G, M, c = sp.symbols('T r theta phi G M c')
gmat = np.diag([1 - 2*G*M/(r*c**2), -1, -r**2, -r**2*sp.sin(teta)**2])
g = es.MetricTensor(gmat, [T, r, teta, phi])
display(g.tensor())
```

```python
Riemann = es.RiemannCurvatureTensor.from_metric(g)
Riemann_Tensor = Riemann.tensor()
Riemann_Tensor
```

```python
display(Riemann_Tensor[1, 0, 0, 1,]), display(Riemann_Tensor[1, 0, 1, 0])
```

```python
display(Riemann_Tensor[1, 0, 0, 1,]), display(Riemann_Tensor[0, 1, 1, 0])
```

```python
gRiemann_hpta = sp.tensorproduct(Riemann.tensor(), g.tensor())
gRiemann_hpta

Rcov = Riemann.change_config('llll')
Rcov
```

```python
ginv = g.change_config('uu')
ginv.tensor()
```

```python
g_Riemann_ricci = sp.tensorproduct(ginv.tensor(), Riemann.tensor())
Ricci_Tensor = sp.simplify(sp.tensorcontraction(g_Riemann_ricci, (0,2), (1,5)))
Ricci_Tensor
```

### Traza de Ricci

```python
gRicci = sp.tensorproduct(ginv.tensor(), Ricci_Tensor)

R = sp.simplify(sp.tensorcontraction(gRicci, (0,2), (1,3)))
R
```
