```python
import numpy as np
import sympy as sp
import einsteinpy.symbolic as es
import matplotlib.pyplot as plt
import pymcel as pc
```

```python
g, x, c = sp.symbols('g x c')
gmat = np.diag([1 + 2*g*x/c**2, -1, -1, -1])
gmat
```

```python
coords = sp.symbols('T x y z')
gmet = es.MetricTensor(gmat, coords)
gtensor = gmet.tensor()
gtensor
```

```python
G = es.ChristoffelSymbols.from_metric(gmet)
G.tensor()
```

```python
xpp = sp.symbols(r"\ddot{T} \ddot{x} \ddot{y} \ddot{z}")
xp = sp.symbols(r"\dot{T} \dot{x} \dot{y} \dot{x^2}")
display(xpp[0])
```

```python
for sigma in range(4):
    ecuacion = 0
    for mu in range(4):
        for nu in range(4):
            ecuacion += -G[sigma, mu, nu] * xp[mu] * xp[nu]
    ecuacion = sp.Eq(xpp[sigma], ecuacion)
    display(ecuacion)
```

```python
def edg_galileo(tau, Ys, g, c):

    T, x, y, z, Ut, Ux, Uy, Uz = Ys
    dT_dtau = Ut
    dx_dtau = Ux
    dy_dtau = Uy
    dz_dtau = Uz
    #Ahora derivadas de U
    dUt_dtau = -2*Ut*Ux*g / (c**2 + 2*g*x)
    dUx_dtau = -g*Ut**2 / c**2
    dUy_dtau = 0
    dUz_dtau = 0

    return [dT_dtau, dx_dtau, dy_dtau, dz_dtau, dUt_dtau, dUx_dtau, dUy_dtau, dUz_dtau]
```

```python
edg_galileo(1, [1, 1, 1, 1, 1, 1, 1, 1], 9.8, 1)
```
